```python
import pandas as pd
from pathlib import Path
data_path = Path("../data/raw") # .. means current se ek folder up jao, data folder ke andar raw folder ke andar jao
```

```python
customers = pd.read_csv(data_path / "customers.csv")
print(customers.shape,
customers.info(),
customers.head())
print(customers['customer_id'].nunique()) # extra likha h, for the number of unique customer_id's in the dataset. It should be equal to the number of rows if there are no duplicates.
print(customers["customer_id"].duplicated().sum()) # to check explicitly duplicate rows k count k lie here was not necessary though
print(customers['customer_id'].describe()) # all the statistical cheezo k lie
customers["city"].value_counts()
customers["signup_date"].min()
customers["signup_date"].max()
pd.to_datetime(customers["signup_date"], errors="coerce").isna().sum() # just for checking if there are any invalid dates in the column.
# If there are any invalid dates, they will be converted to NaT (Not a Time) and counted as NaN values.
customers["signup_date"] = pd.to_datetime(customers["signup_date"]) # ye dataframe k column ko datetime me convert karne k lie hai. not the csv file
customers.info() # transformation k baad validation is very necessary.
customers["signup_date"].min() # ye bhi validate krna hota h ki koi information to chnage ya kuch nai ho gyi
customers["signup_date"].max()

```

```python
categories = pd.read_csv(data_path / "categories.csv")
print(categories.head(),
categories.shape,
categories.info(),
categories["category_id"].duplicated().sum())
```

```python
suppliers = pd.read_csv(data_path / "suppliers.csv")
print(suppliers.head(),
suppliers.shape,
suppliers.info(),
suppliers["supplier_id"].duplicated().sum())
suppliers['country'].value_counts()
```

```python
products = pd.read_csv(data_path / "products.csv")
print("Shape of the dataset:", products.shape)
display(products.head())
products.info()
products["product_id"].duplicated().sum()
print((products["price"] < 0).sum())
(products["price"] == 0).sum()
products["price"].describe()
products["category_id"].isin(categories["category_id"]).all()# check krlia ki sari usme hai ya nia .all() to check for all the 10k rows
products["supplier_id"].isin(suppliers["supplier_id"]).all() # fk established between products and suppliers
```

```python
stores = pd.read_csv(data_path / "stores.csv")
print(stores.head(),
stores.shape,
stores.info(),
stores["store_id"].duplicated().sum())
stores['city'].value_counts()

```

```python
promotions = pd.read_csv(data_path / 'promotions.csv')
print(promotions.head(),
promotions.shape,
promotions.info(),
promotions['promotion_id'].duplicated().sum())
print((promotions["discount"] < 0).sum())
promotions["discount"].describe()
```

```python
orders = pd.read_csv(data_path / "orders.csv")
print(orders.head(),
orders.shape,
orders.info(),
orders["order_id"].duplicated().sum())
print(orders['order_date'].min(),
      orders['order_date'].max())
pd.to_datetime(orders["order_date"], errors="coerce").isna().sum()
orders["order_date"] = pd.to_datetime(orders["order_date"])
orders["customer_id"].isin(customers["customer_id"]).all() #fk established between orders and customers
orders["store_id"].isin(stores["store_id"]).all() #fk established between orders and stores
orders["promotion_id"].isin(promotions["promotion_id"]).all() #fk established between orders and promotions


```

```python
employees = pd.read_csv(data_path / "employees.csv")
print(employees.head(),
employees.shape,
employees.info(),
employees["employee_id"].duplicated().sum())
employees['store_id'].isin(stores['store_id']).all() #fk established between employees and stores
print((employees['salary'] < 0).sum())
print((employees['salary'] == 0).sum())
employees['salary'].describe()
```

```python
order_items = pd.read_csv(data_path / "order_items.csv")
print(order_items.head(),
order_items.shape,
order_items.info(),
order_items["order_item_id"].duplicated().sum())
order_items["order_id"].isin(orders["order_id"]).all() #fk established between order_items and orders
order_items["product_id"].isin(products["product_id"]).all() #fk established between order_items and products
print((order_items['qty'] < 0).sum())
print((order_items['qty'] == 0).sum())
print((order_items['price'] < 0).sum())
print((order_items['price'] == 0).sum())
order_items.describe()
```

```python
payments = pd.read_csv(data_path / "payments.csv")
print(payments.head(),
payments.shape,
payments.info(),
payments['payment_id'].duplicated().sum())
payments['order_id'].isin(orders['order_id']).all() #fk established between payments and orders
print((payments['amount'] < 0).sum())
print((payments['amount'] == 0).sum())
payments['order_id'].duplicated().sum()# very important checking if there are multiple payments for a single order. 1:1 relationship establish krne k lie order_id aur payment_id dono unique hone chahiye. If there are multiple payments for a single order, it could indicate that the order was split into multiple payments or that there was an error in the payment processing.
#If there are, it could indicate that the order was split into multiple payments or that there was an error in the payment processing.
#sirf rows count equal dekh kr nhi 1:1 establish kr skte

```

```python
returns = pd.read_csv(data_path / "returns.csv")
print(returns.head(),
returns.shape,
returns.info(),
'\n',
returns['return_id'].duplicated().sum())
returns['order_item_id'].isin(order_items['order_item_id']).all() #fk established between returns and order_items
print((returns['refund'] < 0).sum())
print((returns['refund'] == 0).sum())
returns["refund"].describe()
```

```python
shipments = pd.read_csv(data_path / "shipments.csv")
print(shipments.head(),
shipments.shape,
shipments.info(),#Every shipment belongs to a valid order, no order has multiple shipment records, and the number of shipments equals the number of orders.
#tab jaakr 1:1 relationship establish kr skte.
'\n',
shipments['shipment_id'].duplicated().sum())
shipments['order_id'].isin(orders['order_id']).all() #fk established between shipments and orders
shipments['order_id'].duplicated().sum() # very important checking if there are multiple shipments for a single order.
#1:1 relationship establish krne k lie order_id aur shipment_id dono unique hone chahiye. 
#If there are multiple shipments for a single order, it could indicate that the order was split into multiple shipments or that there was an error in the shipment processing.
shipments['status'].value_counts() # to check the status of the shipments. 
#It could be shipped, delivered, returned, etc. This will help us understand the current state of the shipments and if there are any issues with them.
```
